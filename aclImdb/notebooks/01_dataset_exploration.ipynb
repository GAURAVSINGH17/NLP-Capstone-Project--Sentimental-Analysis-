We use the Stanford IMDb Large Movie Review Dataset (`aclImdb`).  
Labels come from folder names (`pos` = positive, `neg` = negative).
No neutral class exists in this dataset.

```python
import os, sys
import pandas as pd
import numpy as np

# Point to the project root (aclImdb/)
PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
sys.path.insert(0, PROJECT_ROOT)

from src.data_loader import load_imdb_dataset, get_train_test

print('Project root:', PROJECT_ROOT)
```

## 1. Load the dataset

```python
df = load_imdb_dataset(PROJECT_ROOT)
print(f'Total records loaded: {len(df):,}')
df.head(3)
```

## 2. Dataset summary

```python
print('Shape:', df.shape)
print('\nColumns:', df.columns.tolist())
print('\nSplit distribution:')
print(df['split'].value_counts())
print('\nSentiment distribution:')
print(df['sentiment'].value_counts())
print('\nSentiment by split:')
print(df.groupby(['split', 'sentiment']).size().unstack())
```

## 3. Sample reviews

```python
train_df, test_df = get_train_test(df)

print('=== POSITIVE REVIEW SAMPLE ===')
print(train_df[train_df['sentiment'] == 'positive']['review'].iloc[0][:500])
print('\n=== NEGATIVE REVIEW SAMPLE ===')
print(train_df[train_df['sentiment'] == 'negative']['review'].iloc[0][:500])
```

## 4. Data cleaning checks
We check for: missing text, empty/whitespace reviews, duplicates, encoding issues.

```python
# Missing values
print('Missing values per column:')
print(df.isnull().sum())

# Empty or whitespace-only reviews
empty_mask = df['review'].str.strip().eq('')
print(f'\nEmpty/whitespace-only reviews: {empty_mask.sum()}')

# Duplicates
dup_count = df.duplicated(subset=['review']).sum()
print(f'Duplicate reviews: {dup_count}')

# Review length stats
df['review_len_chars'] = df['review'].str.len()
df['review_len_words'] = df['review'].str.split().str.len()
print('\nReview length (words):')
print(df['review_len_words'].describe().round(1))
```

```python
# Check for unexpected label values
print('Unique sentiment values:', df['sentiment'].unique())
print('Unique label values:', df['label'].unique())
print('\nLabel -> sentiment mapping:')
print(df.groupby(['label','sentiment']).size())
```

### Cleaning conclusions
- No missing review text detected.
- No empty/whitespace-only reviews.
- Duplicate count reported above (the IMDb dataset has no known duplicates in labelled splits).
- Both positive and negative labels are confirmed.
- Dataset is clean and ready for text preprocessing.
