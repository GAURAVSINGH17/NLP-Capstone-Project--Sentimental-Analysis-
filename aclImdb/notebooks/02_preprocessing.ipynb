We demonstrate what `src/preprocessing.py` does and show before/after examples.

```python
import os, sys
import pandas as pd

PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
sys.path.insert(0, PROJECT_ROOT)

from src.data_loader import load_imdb_dataset, get_train_test
from src.preprocessing import clean_review, apply_cleaning, NEGATION_WORDS

df = load_imdb_dataset(PROJECT_ROOT)
train_df, test_df = get_train_test(df)
print('Loaded', len(train_df), 'training reviews')
```

## Preprocessing steps applied
1. Decode HTML entities
2. Remove HTML tags (e.g. `<br />`)
3. Lowercase
4. Keep letters and apostrophes only (preserves contractions)
5. Collapse whitespace
6. (Optional) Stopword removal — negation words always kept

```python
# Before/after examples
examples = [
    'The movie was <b>fantastic</b>! Acting &amp; story were top-notch.',
    'I did NOT enjoy this film at all.<br />It was absolutely terrible.',
    'Never going to watch this again. Worst film EVER made.',
]

for raw in examples:
    cleaned = clean_review(raw)
    print(f'RAW     : {raw}')
    print(f'CLEANED : {cleaned}')
    print()
```

```python
# Negation words that are explicitly preserved
print('Preserved negation words:')
print(sorted(NEGATION_WORDS))
```

```python
# Apply to train data and show comparison
train_df['clean_review'] = apply_cleaning(train_df['review'])

print('=== Original review (first 300 chars) ===')
print(train_df['review'].iloc[0][:300])
print()
print('=== Cleaned review (first 300 chars) ===')
print(train_df['clean_review'].iloc[0][:300])
```

## Note on TF-IDF and preprocessing
For the main model pipeline, `TfidfVectorizer` performs tokenisation internally.
The `clean_review` function handles HTML removal and lowercasing, which are
not done by `TfidfVectorizer` by default.  Stemming and lemmatisation are
**not** applied — they can harm performance for a TF-IDF + linear classifier baseline.
